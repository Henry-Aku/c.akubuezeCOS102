```python
import tkinter as tk
from tkinter import messagebox
import random
from datetime import date
```

```python
class Employee:
    def __init__(self):
        self.employees = [
            "Mary Evans", "Eyo Ishan", "Durojaiye Dare", "Adams Ali",
            "Andrew Ugwu", "Stella Mankinde", "Jane Akibo", "Ago James",
            "Michell Taiwo", "Abraham Jones", "Nicole Anide", "Kosi Korso",
            "Adele Martins", "Emmanuel Ojo", "Ajayi Fatima"
        ]
        self.tasks = [
            "Loading", "Transporting", "Reviewing Orders",
            "Customer Service", "Delivering Items"
        ]
        self.attendance_log = {}

    def check_employee(self, name):
        """Check if the name exists in the employee list (case-insensitive)."""
        for emp in self.employees:
            if emp.lower() == name.strip().lower():
                return emp  # Return the correctly-cased name
        return None

    def take_attendance(self, name):
        """Record attendance for the employee for today."""
        today = str(date.today())
        if today not in self.attendance_log:
            self.attendance_log[today] = []
        if name not in self.attendance_log[today]:
            self.attendance_log[today].append(name)
            return True  # First check-in today
        return False  # Already checked in

    def assign_task(self):
        """Randomly assign a task from the task list."""
        index = random.randint(0, len(self.tasks) - 1)
        return self.tasks[index]

    def refuse_access(self):
        """Return a polite refusal message."""
        return "Access Denied. You are not recognized as an employee of this organization."


# ─── GUI Application ───────────────────────────────────────────────────────────

class App:
    def __init__(self, root):
        self.root = root
        self.root.title("Mrs. Jane's Delivery — Employee Attendance System")
        self.root.geometry("520x420")
        self.root.resizable(False, False)
        self.root.configure(bg="#1e1e2e")

        self.employee_mgr = Employee()
        self._build_ui()

    def _build_ui(self):
        # Title
        tk.Label(
            self.root, text="🚚  Employee Check-In System",
            font=("Helvetica", 16, "bold"),
            bg="#1e1e2e", fg="#cdd6f4"
        ).pack(pady=(24, 4))

        tk.Label(
            self.root, text="Mrs. Jane's Delivery Business",
            font=("Helvetica", 10), bg="#1e1e2e", fg="#a6adc8"
        ).pack()

        # Separator
        tk.Frame(self.root, bg="#45475a", height=1).pack(fill="x", padx=30, pady=16)

        # Name entry
        tk.Label(
            self.root, text="Enter Employee Full Name:",
            font=("Helvetica", 11), bg="#1e1e2e", fg="#cdd6f4"
        ).pack()

        self.name_var = tk.StringVar()
        entry = tk.Entry(
            self.root, textvariable=self.name_var,
            font=("Helvetica", 12), width=30,
            bg="#313244", fg="#cdd6f4",
            insertbackground="#cdd6f4",
            relief="flat", bd=6
        )
        entry.pack(pady=10)
        entry.bind("<Return>", lambda e: self._process())

        # Check-in button
        tk.Button(
            self.root, text="  Check In  ",
            font=("Helvetica", 11, "bold"),
            bg="#89b4fa", fg="#1e1e2e",
            activebackground="#74c7ec",
            relief="flat", bd=0, padx=16, pady=6,
            cursor="hand2",
            command=self._process
        ).pack(pady=6)

        # Result box
        tk.Frame(self.root, bg="#45475a", height=1).pack(fill="x", padx=30, pady=16)

        self.result_frame = tk.Frame(self.root, bg="#1e1e2e")
        self.result_frame.pack(fill="both", padx=30)

        self.result_label = tk.Label(
            self.result_frame, text="",
            font=("Helvetica", 11), bg="#1e1e2e", fg="#a6e3a1",
            wraplength=440, justify="left"
        )
        self.result_label.pack(anchor="w")

        # Reset button (hidden initially)
        self.reset_btn = tk.Button(
            self.root, text="Reset",
            font=("Helvetica", 9),
            bg="#45475a", fg="#cdd6f4",
            activebackground="#585b70",
            relief="flat", bd=0, padx=10, pady=4,
            cursor="hand2",
            command=self._reset
        )

    def _process(self):
        name_input = self.name_var.get().strip()
        if not name_input:
            messagebox.showwarning("Input Required", "Please enter a name.")
            return

        matched_name = self.employee_mgr.check_employee(name_input)

        if matched_name:
            already_in = not self.employee_mgr.take_attendance(matched_name)
            task = self.employee_mgr.assign_task()
            today = str(date.today())

            if already_in:
                msg = (
                    f"✅  Welcome back, {matched_name}!\n\n"
                    f"📋  Attendance: Already recorded for {today}.\n"
                    f"🔧  Task for today: {task}"
                )
            else:
                msg = (
                    f"✅  Welcome, {matched_name}!\n\n"
                    f"📋  Attendance recorded for {today}.\n"
                    f"🔧  Task assigned for today: {task}"
                )
            self.result_label.config(text=msg, fg="#a6e3a1")
        else:
            msg = f"❌  {self.employee_mgr.refuse_access()}\n\nName entered: \"{name_input}\""
            self.result_label.config(text=msg, fg="#f38ba8")

        self.reset_btn.pack(pady=(12, 0))

    def _reset(self):
        self.name_var.set("")
        self.result_label.config(text="")
        self.reset_btn.pack_forget()


if __name__ == "__main__":
    root = tk.Tk()
    app = App(root)
    root.mainloop()
```
