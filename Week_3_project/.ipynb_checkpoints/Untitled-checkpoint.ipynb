```python
import tkinter as tk
from tkinter import messagebox, ttk
import pandas as pd

class GIGVerifyApp:
    def __init__(self, root):
        self.root = root
        self.root.title("GIG Logistics - Employee Verification")
        self.root.geometry("500x400")

        # Load the dataset
        try:
            self.df = pd.read_csv('GIG-logistics.csv')
            # Clean column names in case of trailing spaces
            self.df.columns = self.df.columns.str.strip()
        except FileNotFoundError:
            messagebox.showerror("Error", "GIG-logistics.csv not found!")
            self.root.destroy()

        # UI Elements
        tk.Label(root, text="Employee Name:", font=('Arial', 10, 'bold')).pack(pady=5)
        self.name_entry = tk.Entry(root, width=40)
        self.name_entry.pack(pady=5)

        tk.Label(root, text="Department:", font=('Arial', 10, 'bold')).pack(pady=5)
        self.dept_entry = tk.Entry(root, width=40)
        self.dept_entry.pack(pady=5)

        self.verify_btn = tk.Button(root, text="Verify Identity", command=self.verify_employee, bg="#2196F3", fg="white")
        self.verify_btn.pack(pady=20)

        # Results area
        self.result_label = tk.Label(root, text="", wraplength=400, justify="left")
        self.result_label.pack(pady=10)

    def verify_employee(self):
        input_name = self.name_entry.get().strip()
        input_dept = self.dept_entry.get().strip()

        # Filter the dataframe (Case-insensitive check)
        user_match = self.df[
            (self.df['Name'].str.contains(input_name, case=False, na=False)) & 
            (self.df['Department'].str.contains(input_dept, case=False, na=False))
        ]

        if not user_match.empty:
            full_name = user_match.iloc[0]['Name']
            dept_name = user_match.iloc[0]['Department']
            
            # Welcome message
            welcome_text = f"Welcome, {full_name}!\nVerification Successful.\n\n"
            
            # Get other members of the department
            colleagues = self.df[self.df['Department'] == dept_name]['Name'].tolist()
            if full_name in colleagues: colleagues.remove(full_name)
            
            colleagues_str = ", ".join(colleagues) if colleagues else "None"
            welcome_text += f"Your colleagues in {dept_name}: \n{colleagues_str}"
            
            self.result_label.config(text=welcome_text, fg="green")
            messagebox.showinfo("Success", f"Identity Verified for {full_name}")
        else:
            self.result_label.config(text="Identity not found. Please check details or contact HR.", fg="red")
            messagebox.showwarning("Access Denied", "Employee does not exist in our records.")

if __name__ == "__main__":
    root = tk.Tk()
    app = GIGVerifyApp(root)
    root.mainloop()
```
